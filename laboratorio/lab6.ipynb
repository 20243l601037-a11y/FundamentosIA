```python
#Laboratorio 6: Grafos
#Descripcion: Encuentra la ruta
#Situacion: Un robot movil inteligente se encuentra en A y necesita llegar a G.
```

```python
#Librerias utilizada para graficar
import matplotlib.pyplot as ptl
#Libreria que me permite trabajar con estructuras complejas como grafos
import networkx as nx
```

```python
#Creamos una instancia de la clase graph
#Graph es una coleccion de nodos
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizando el método add_edges_from

grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("C","G"),
        ("D","F"),
        ("G","E"),
        ("G","J"),
        ("E","F"),
        ("F","H"),
        ("H","I"),
        ("I","J")
    ]
)
```

```python
# Mostrar el grafo
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#e44b8d"
)
plt.show()
```

```python
#Mostrar las rutas posibles del grafo
#SOURCE = De donde inicia
#Target = El objetivo
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas:")
for i,ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Mostrar el total de pasos de cada ruta
paso1 = len(rutas[0]) - 1
print("Pasos de la ruta 1: ", paso1)
paso2 = len(rutas[1]) - 1
print("Pasos de la ruta 2: ", paso2)
paso3 = len(rutas[2]) - 1
print("Pasos de la ruta 3: ", paso3)
paso4 = len(rutas[3]) - 1
print("Pasos de la ruta 4: ", paso4)
```

```python
# Aplicamos la lógica proposicional
P = paso1 < paso2 and paso1 < paso3 and paso1 < paso4 # Evalúa ruta 1
Q = paso2 < paso1 and paso2 < paso3 and paso2 < paso4 # Evalúa ruta 2
R = paso3 < paso1 and paso3 < paso2 and paso3 < paso4 # Evalúa ruta 3
S = paso4 < paso1 and paso4 < paso2 and paso4 < paso3 # Evalúa ruta 4

if P:
    print("La ruta 1 es la mejor opción.")
elif Q:
    print("La ruta 2 es la mejor opción.")
elif R:
    print("La ruta 3 es la mejor opción.")
elif S:
    print("La ruta 4 es la mejor opción.")
else:
    print("Hay empate entre rutas.")
```

```python
import networkx as nx
import matplotlib.pyplot as plt
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("A", "B"),
        ("A", "C"),
        ("B", "D"),
        ("C", "G"),
        ("D", "F"),
        ("E", "F"),
        ("E", "G"),
        ("F", "H"),
        ("G", "J"),
        ("H", "I"),
        ("I", "J")
    ]
)


grafo.remove_edge("G", "J")

nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#e44b8d"
)
plt.show()
```

```python
# Obtener las nuevas rutas 
rutas = list(nx.all_simple_paths(grafo, source="A", target="J"))

# Imprimir las rutas 
print("Rutas encontradas tras el bloqueo:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Mostrar el total de pasos de las rutas restantes
paso1 = len(rutas[0]) - 1
print("Pasos de la ruta 1: ", paso1)

paso2 = len(rutas[1]) - 1
print("Pasos de la ruta 2: ", paso2)
```

```python
# Aplicamos la lógica proposicional
P = paso1 < paso2 # P: La ruta 1 es la mejor opción
Q = paso2 < paso1 # Q: La ruta 2 es la mejor opción

if P:
    print("La ruta 1 es la mejor opción tras el bloqueo.")
elif Q:
    print("La ruta 2 es la mejor opción tras el bloqueo.")
else:
    print("Ambas rutas son igual de largas.")
```
