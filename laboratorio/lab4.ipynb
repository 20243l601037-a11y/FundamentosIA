### Laboratorio 4: Propociones
### **objetivo:** Un sistema inteligente supervisa una red empresarial y 
### determina si existe una condificion ctritica

**Fecha:** 7 de septiembre

```python
#Propociones 
P: El servidor principal está disponible.
Q: La latencia supera el límite establecido (latencia alta).
R: La pérdida de paquetes supera el límite permitido (pérdida de paquetes alta).
S: El tráfico de red es anormalmente alto.
T: El servidor de respaldo está disponible.
U: El sistema de monitoreo está activo.
```

```python
#Definición de variables
servidorPrincipal = False # El servidor principal no está disponible
latenciaAlta = True # Existe latencia alta
perdidaPaquetes = True # Existe pérdida de paquetes alta
traficoAlto = True # El tráfico es anormalmente alto
servidorRespaldo = False # El servidor de respaldo no está disponible
sistemaMonitoreo = True # El sistema de monitoreo está activo
```

```python
condicionCritica = (servidorPrincipal==False and latenciaAlta==True and perdidaPaquetes==True and traficoAlto==True and servidorRespaldo==False and sistemaMonitoreo==True)
```

```python
print("¿Existe una condición crítica de servicio?", condicionCritica)
```

```python
print (" == Sistema Inteligente de Red Empresarial ==")
print ("Development by Antonio")

servidorPrincipal = input("El servidor principal esta disponible (s/n): ").lower() == "s"
latenciaAlta = input("Existe latencia alta (s/n): ").lower() == "s"
perdidaPaquetes = input("Existe perdida de paquetes alta (s/n): ").lower() == "s"
traficoAlto = input("El trafico es anormalmente alto (s/n): ").lower() == "s"
servidorRespaldo = input("El servidor de respaldo esta disponible (s/n): ").lower() == "s"
sistemaMonitoreo = input("El sistema de monitoreo esta activo (s/n): ").lower() == "s"

condicionCritica = (servidorPrincipal==False and latenciaAlta==True and perdidaPaquetes==True and traficoAlto==True and servidorRespaldo==False and sistemaMonitoreo==True)

print ("\nDecision del sistema inteligente")
if condicionCritica == True:
    print("¡ALERTA CRÍTICA! Se cumplen todas las condiciones de fallo. Revisa la red inmediatamente.")
else:
    print("Estado normal. No existe una condición crítica de servicio en este momento.")
```
