```python
#Grafo 2
#importar las librerías
#librería utilizada para graficar
import matplotlib.pyplot as plt
#librería que me permite trabajar con estructuras complejas, como Grafos
import networkx as nx
```

```python
# Creamos el grafo
grafo = nx.DiGraph()

grafo.add_weighted_edges_from([
    ("A", "B", 4),
    ("A", "C", 7),
    ("B", "D", 3),
    ("C", "D", 2),
    ("F", "G", 4),
    ("D", "G", 8),
    ("D", "E", 2),
    ("B", "E", 6),
    ("E", "G", 3),
    ("C", "F", 5)
])
```

```python
# Imprimimos el grafo
pos = nx.spring_layout(grafo, seed=10)

nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
# Mostramos las rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="G"
    )
)

# Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Calculamos el costo
def costo_ruta(grafo, ruta):
    costo = 0
    for i in range(len(ruta) - 1):
        costo += grafo[ruta[i]][ruta[i + 1]]["weight"]
    return costo

# Imprimimos el costo
for i, ruta in enumerate(rutas, start=1):
    costo = costo_ruta(grafo, ruta)
    print(f"Ruta {i}: {ruta} -> {costo} unidades")
```

```python
# Elegir la ruta más corta

costos = [ costo_ruta (grafo, ruta)
          for ruta in rutas
         ]

mejor_ruta = costos.index ( min ( costos) )
```

```python
# Imprimir la mejor ruta

print(f"La mejor ruta es: {mejor_ruta + 1}")
print(f"El camino costo es: {costos[mejor_ruta]}")
```
