```python
#importar las librerías
#librería utilizada para graficar
import matplotlib.pyplot as plt
#librería que me permite trabajar con estructuras complejas, como Grafos
import networkx as nx
```

```python
#Creamos una instancia de la clase graph
#Graph es una coleccion de nodos
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizando el metodo add_edges_from

grafo.add_edges_from(
    [
        ("Servidor_1", "RouterA"),
        ("RouterA", "SwitchA"),
        ("RouterA", "SwitchB"),
        ("SwitchA", "PC1"),
        ("SwitchA", "ImpresoraA"),
        ("SwitchB", "PC2"),
        ("SwitchB", "ImpresoraA")
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels=True,
    node_size=2000,
    node_color="#e44b8d"
)

plt.show()
```
