```python
#importar las librerías
#librería utilizada para graficar
import networkx as nx
#librería que me permite trabajar con estructuras complejas, como Grafos
import matplotlib.pyplot as plt
```

```python
# Agregar las rutas con sus pesos
grafo = nx.DiGraph()

grafo.add_weighted_edges_from([
    ("A", "B", 8),
    ("A", "E", 4),
    ("A", "D", 5),

    ("B", "C", 3),
    ("B", "F", 4),

    ("C", "G", 2),
    ("C", "F", 9),

    ("F", "E", 3),
    ("F", "K", 8),

    ("D", "E", 3),
    ("D", "H", 6),

    ("E", "I", 5),

    ("H", "I", 4),
    ("H", "M", 2),

    ("I", "J", 4),
    ("I", "N", 3),
    ("I", "M", 1),

    ("M", "N", 1),

    ("N", "O", 2),

    ("J", "O", 9),

    ("G", "K", 3),
    ("G", "L", 2),

    ("L", "K", 5),
    ("L", "P", 3),

    ("K", "J", 6),

    # OJO: esta flecha va de P hacia K
    ("P", "K", 1),

    ("O", "P", 12)
])
```

```python
# Imprimimos el grafo
pos = nx.spring_layout(grafo, seed=10)

nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)

plt.show()
```

```python
# Mostramos las rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="P"
    )
)

# Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Calculamos el costo
def costo_ruta(grafo, ruta):
    costo = 0
    for i in range(len(ruta) - 1):
        costo += grafo[ruta[i]][ruta[i + 1]]["weight"]
    return costo

# Imprimimos el costo
for i, ruta in enumerate(rutas, start=1):
    costo = costo_ruta(grafo, ruta)
    print(f"Ruta {i}: {ruta} -> {costo} unidades")
```

```python
# Utilizamos algoritmo para obtener la ruta más corta

ruta_mas_corta = nx.dijkstra_path(grafo, source="A", target="P", weight='weight')

tiempo_total = nx.dijkstra_path_length(grafo, source="A", target="P", weight="weight")
```

```python
# Imprimir

print(f"La ruta más corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```

```python
# Utilizamos algoritmo para obtener la ruta crítica

rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="P"
    )
)

ruta_critica = max(
    rutas,
    key=lambda ruta: nx.path_weight(grafo, ruta, weight="weight")
)

tiempo_critico = nx.path_weight(
    grafo,
    ruta_critica,
    weight="weight"
)
```

```python
# Imprimir

print(f"La ruta crítica es: {ruta_critica}")
print(f"El tiempo total es: {tiempo_critico} minutos")
```
